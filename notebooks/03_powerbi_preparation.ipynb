```python
import pandas as pd
df = pd.read_csv('/content/Cleaned_Youtube_With_Sentiment.csv', on_bad_lines='skip', engine='python')
```

```python
import os

time_series = df.groupby(['trending_date','country'])['views'].sum().reset_index()
os.makedirs('data/sql_output', exist_ok=True)
time_series.to_csv('data/sql_output/time_series_views.csv', index=False)
```

```python
cat_avg = df.groupby(['country','category_name'])['views'].mean().reset_index()
cat_avg.to_csv('data/sql_output/category_avg_views.csv', index=False)
```

```python
sent_summary = df.groupby(['country','sentiment_label'])['video_id'].count().reset_index()
sent_summary.to_csv('data/sql_output/sentiment_distribution.csv', index=False)
```

```python
duration = df.groupby(['video_id','country'])['trending_date'].count().reset_index()
duration.rename(columns={'trending_date':'days_trending'}, inplace=True)
duration.to_csv('data/sql_output/trending_duration.csv', index=False)
```
