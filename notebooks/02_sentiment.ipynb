```python
import pandas as pd

df = pd.read_csv('/content/Cleaned_Youtube.csv')
df.head()
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
sia = SentimentIntensityAnalyzer()
```

```python
df['title_sentiment'] = df['title'].astype(str).apply(
    lambda x: sia.polarity_scores(x)['compound']
)
```

```python
df[['title', 'title_sentiment']].head()
```

```python
def label_sent(score):
    if score >= 0.05:
        return 'Positive'
    elif score <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['sentiment_label'] = df['title_sentiment'].apply(label_sent)
```

```python
df[['title', 'title_sentiment', 'sentiment_label']].head()
```

```python
df.to_csv('/content/Cleaned_Youtube_With_Sentiment.csv', index=False)
```
